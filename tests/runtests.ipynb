```python
import os

def get_user_workspace_path():
    user_workspace_path = ""
    try:
        current_path = os.getcwd()

        # Split the path to extract up to the user's email directory
        path_parts = current_path.split("/")
        user_workspace_path = "/".join(path_parts[:4]) + "/"
    finally:
        return user_workspace_path
```

```python
import pytest
import sys

# Define the root repository directory
repo_root = get_user_workspace_path() + "tfnsw-data-ai_platform"

if repo_root not in sys.path:
    sys.path.insert(0, repo_root)

# Disable writing bytecode (.pyc) files to the workspace mount
sys.dont_write_bytecode = True

# Run pytest pointing cache_dir to local /tmp
pytest.main([
    "-v",
    "-p", "no:cacheprovider",
    "-o", "cache_dir=/tmp/.pytest_cache",
    os.path.join(repo_root, "tests")
])
```
