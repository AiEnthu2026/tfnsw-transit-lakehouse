```python
import os
import sys
import importlib
import pytest

# Find the repo root: walk up from this notebook's folder until a "tests" folder exists
root = os.getcwd()
while root != "/" and not os.path.isdir(os.path.join(root, "tests")):
    root = os.path.dirname(root)
repo_root = root

# Disable writing bytecode (.pyc) files to the workspace mount
sys.dont_write_bytecode = True
# pick up files the bundle just deployed
importlib.invalidate_caches()
if repo_root not in sys.path:
    sys.path.insert(0, repo_root)
```

```python
# Job parameters: test tier to run, and which catalog the tests should use
dbutils.widgets.text("test_path", "unit")
dbutils.widgets.text("catalog_name", "dev")
test_path = dbutils.widgets.get("test_path")
os.environ["CATALOG_NAME"] = dbutils.widgets.get("catalog_name")

target_dir = os.path.join(repo_root, "tests", test_path)

exit_code = pytest.main([
    "-v",
    "--import-mode=importlib",
    "-p", "no:cacheprovider",
    "-o", "cache_dir=/tmp/.pytest_cache",
    target_dir
])

if exit_code != 0:
    raise Exception(f"Test suite failed with pytest exit code {exit_code}")
```
